## Streaming 
#### Most models can stream their output while it is being generated. By displaying the output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yiels output chunks as they are produced. You can use loop to process each chunk in real-time.

```python
groq_model.invoke("write a 200 words paragraph on Artificial Intelligence")
```

```python
for chunk in chat_model.stream("write a 200 words paragraph on Artificial Intelligence"):
    print(chunk.text, end="|", flush=True)
```

## Batching

```python
responses = gemini_model.batch([
    "why do parrots talk?",
    "how to airplanes fly?",
    "what is quantum computing?"],
    config={
        'max_concurrency': 3, #limit to 3 parallel requests
    }
)

for response in responses:
    print(response)
```
